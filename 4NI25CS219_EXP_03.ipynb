```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/student_data.csv")
print("Dataset shape:",df.shape)
```

```python
print(df.head() )
```

```python
print(df.info())
```

```python
print(df.describe())
```

1 -> Histogram -- distribution of final grade

```python
plt.figure(figsize=(6,4))
sns.histplot(df["G3"],bins=20,kde=True,color="steelblue")
plt.title ("Distribution of Final Grade (G3)")
plt.xlabel("Final Grade")
plt.ylabel("Frequency")
plt.show()
```

**2 --> BoxPLot**
-grade spread across study time levels .

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="studytime",y="G3",data=df,palette="Set2")
plt.title ("Grade Spread Across Study Time Levels")
plt.xlabel("Study Time(1=low ,4=high)")
plt.ylabel("Final Grade")
plt.show()
```

**3. Scatter plot **--> relationship between two grades (b/w two data columns)

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x="G1",y="G3",data=df,hue="sex",palette="Set2")
plt.title ("First Period Grade vs Final Grade")
plt.xlabel("G1(Period 1 Grade)")
plt.ylabel("G3 (Final Grade)")
plt.show()
```

**4.Pair Plot** --> pairwise relatioships across the key numeric columns

```python
sns.pairplot(df[["age","studytime","G1","G2","G3","sex"]],hue="sex",palette="Set2")
plt.suptitle("Pairwise Relationship:",y=1.02)
plt.show()
```

5. Correlational Heatmap --> strength of relationships between numeric features

```python
numeric_cols=["age","studytime","failures","absences","G1","G2","G3"]
plt.figure(figsize=(8,6))
sns.heatmap(df[numeric_cols].corr(),annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlational Heatmap")
plt.show()
```
